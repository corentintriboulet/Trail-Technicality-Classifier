```python
import numpy as np
import pandas as pd
```

```python
raw_parquet = pd.read_parquet('../data/raw/reunion_segments.parquet')
df_parquet = pd.DataFrame(raw_parquet)
df_parquet = df_parquet.rename(columns={"id": "segment_id"})
df_parquet_ride = df_parquet[df_parquet['activity_type'] == 'Run']
```

### By names = "route" or "rte" or "rn"

```python
T1_keywords = ["route ", "rte ", "rn ", "rd ", "rue "]
excluded_keywords = ["forestiere ", "forestière ", "forest ", "foret ", "forêt "]

df_parquet_ride = df_parquet_ride[
    df_parquet_ride['name'].str.contains('|'.join(T1_keywords), regex=True, na=False, case=False)
    & ~df_parquet_ride['name'].str.contains('|'.join(excluded_keywords), regex=True, na=False, case=False)
].copy()
```

```python
df_ride_T1 = df_parquet_ride[['segment_id']].copy()
df_ride_T1.loc[:, "technicality"] = np.ones(len(df_ride_T1), dtype=int)
df_ride_T1.to_csv('../data/processed/T1_run_road_naming_ride.csv', index=False)
```
